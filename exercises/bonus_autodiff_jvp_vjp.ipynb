# Function Signatures

You should feel somewhat comfortable with function signatures at this point, so let's dive in! As in the [previous material on autodiff](./exe_07_grad.ipynb), we introduce some notation:

Let `T a` denote the tangent space of `a` and let `CT a` denote the cotangent space of `a`, where they are duals of one another. We can think of the tangent space being the space on which we can "move" i.e. the small perturbations we can make, while the cotangent space describes how to measure those movements.

`jvp :: (a -> b) -> a -> T a -> (b, T b)`

`vjp :: (a -> b) -> a -> (b, CT b -> CT a)`


As we can see, the two are intimately tied together, but also rather complex and can involve differential geometry, which is why this was left for a supplementary section.
